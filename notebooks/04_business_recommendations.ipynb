Синтетические наблюдательные данные. Ноутбук не строит новых моделей и не утверждает причинность. Сценарии — `illustrative_scenario_not_forecast`.

```python
from pathlib import Path

import pandas as pd

from delivery_pulse.recommendations import RecommendationConfig, run_recommendations

root = Path.cwd()
database = root / "data" / "processed" / "delivery_pulse.duckdb"
hypothesis_dir = root / "reports" / "hypotheses"
output_dir = root / "reports" / "recommendations"
result = run_recommendations(
    RecommendationConfig(database, hypothesis_dir, output_dir, force=True)
)
result.elapsed_seconds
```

## Evidence matrix и приоритеты

```python
scores = pd.read_csv(result.output_paths["scores"])
scores[
    ["recommendation_id", "evidence_level", "priority", "action_type", "priority_score"]
]
```
